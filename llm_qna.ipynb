```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.7,
)
```

```python
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
graph = StateGraph(LLMState)
```

```python
def ask_llm(state: LLMState) -> LLMState:
    q = state['question']
    prompt = f"Answer the following question: \n {q}"
    result = llm.invoke(prompt)
    state['answer'] = result.content
    return state
```

```python
graph.add_node('ask_llm', ask_llm)

graph.add_edge(START, 'ask_llm')
graph.add_edge('ask_llm', END)
```

```python
workflow = graph.compile()
```

```python
input_state = {
    'question': "What is the distance between sun and moon?"
}

final_state = workflow.invoke(input_state)

print(final_state['answer'])
```
